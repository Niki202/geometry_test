```python
from sympy import symbols, init_printing, Integral, sin
from IPython.display import display, Markdown

# Включаем автоматический вывод в LaTeX
init_printing(use_latex=True)

x = symbols('x')
expr = Integral(sin(x), x)

# Теперь print() автоматически выведет LaTeX-код (в консоли)
# или отрендерит красивую картинку (в Jupyter Notebook)
display(expr) 
display(Markdown(r'$$\frac{3}{5}$$'))
```
